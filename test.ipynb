```python
import os
import json
import spacy
from sentence_transformers import SentenceTransformer
from bertopic import BERTopic

# Load the spaCy language model
nlp = spacy.load("en_core_web_sm")

# 1. Your documents (captions)

# Path to the folder containing the subtitle chunks
folder_path = 'subtitle_chunks_json'

# List to store all the texts
docs = []

# Loop through all JSON files in the folder
for filename in os.listdir(folder_path):
    if filename.endswith('.json'):
        file_path = os.path.join(folder_path, filename)
        
        # Load JSON data
        with open(file_path, 'r') as file:
            subtitle_data = json.load(file)
            
            # Extract 'text' from each chunk and process with spaCy
            for chunk in subtitle_data:
                text = chunk['text']
                
                # Process the text using spaCy
                doc = nlp(text)
                
                # Remove stop words, non-alphabetic tokens, and lemmatize the words
                cleaned_text = ' '.join([token.lemma_ for token in doc if not token.is_stop and token.is_alpha])

                docs.append(cleaned_text)

# 2. Generate embeddings using SentenceTransformers
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = embedding_model.encode(docs, show_progress_bar=True)

# 3. Fit BERTopic
topic_model = BERTopic(verbose=True)
topics, probs = topic_model.fit_transform(docs, embeddings)

# 4. View the topic info
print(topic_model.get_topic_info())

# Optional: View top words in a topic
print("\nTop words for Topic 0:")
print(topic_model.get_topic(0))
```
